```python
#Print a 5x5 grid displaying multiplication tables for numbers 1 to 5

for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row*col:3}", end=' ')
    print()

```

**nested for loop practice**

```python
for i in range(5):
  for j in range(5):
      print("*",end="")
  print()
```

```python
for i in range(6):
  for j in range(4):
    print(i,end=" ")
  print()
```

```python
for row in range(0,5):
    for column in range(3):
       if row == 3 and column==2 or row == 2 and column == 2:
          print("*", end=" ")
       else:
         print(".",end=" ")
    print()
```

```python
for i in range(0,7):
  for j in range(6):
    if (i==6 and j==5) or (i==2 and j==2):
        print("*",end=" ")
    else:
       print(".",end=" ")
  print()
```
