```python
# question2
#Task 1: Create a DataFrame with missing salary values
import pandas as pd
import numpy as np

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Ravi", "Priya", "Amit", "Sneha", "Kiran", "Divya"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Age": [25, 30, 28, 35, 27, 32],
    "Salary": [50000, np.nan, 60000, 70000, np.nan, 80000]
}

df = pd.DataFrame(data)
print(df)
```

```python
# task-2 Count missing values in each column
print(df.isnull().sum())
```

```python
# task-3 Fill missing salaries with the median salary of each department
df["Salary"] = df["Salary"].fillna(
    df.groupby("Department")["Salary"]
      .transform("median")
)

print(df)
```

```python
#Task-4: Calculate annual bonus and average salary per department
df["Annual_Bonus"] = df["Salary"] * 0.10

average_salary = df.groupby("Department")["Salary"].mean()

print("DataFrame with Annual Bonus:")
print(df)

print("\nAverage Salary by Department:")
print(average_salary)
```

```python
#Q4. Pandas Filtering & Conditional Selection
#Task 1: Create a DataFrame
import pandas as pd

data = {
    "Product": ["Laptop", "Mouse", "Monitor", "Keyboard", "Printer", "Tablet"],
    "Category": ["Electronics", "Accessories", "Electronics",
                 "Accessories", "Electronics", "Electronics"],
    "Price": [60000, 400, 12000, 800, 15000, 25000],
    "Stock": [15, 100, 30, 10, 45, 60]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Task 2: Filter products where Price > 500 and Stock < 50
filtered_df = df[(df["Price"] > 500) & (df["Stock"] < 50)]

print(filtered_df)
```

```python
#Task 3–4: Add Restock_Needed and sort by Price descending
df["Restock_Needed"] = df["Stock"] < 20

df = df.sort_values(by="Price", ascending=False)

print(df)
```

```python
# Q5. Vectorized Temperature & Math Operations
#Task 1–2: Generate Celsius temperatures and convert to Fahrenheit
import numpy as np

# Generate 10 random temperatures between 15 and 40 Celsius
celsius = np.random.uniform(15.0, 40.0, size=10)

# Convert all temperatures to Fahrenheit
fahrenheit = (celsius * 9 / 5) + 32

print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
```

```python
#Task 3: Calculate square roots of numbers from 1 to 10
arr = np.arange(1, 11)

sqrt_arr = np.sqrt(arr)

print(sqrt_arr)
```

```python
#Task 4: Find the index of the maximum Fahrenheit temperature
max_index = np.argmax(fahrenheit)

print("Index of maximum temperature:", max_index)
print("Maximum Fahrenheit temperature:", fahrenheit[max_index])
```
